**AUTOMATED GENERATED FINANCIAL REPORT**
==============================

**Function:** This script generates basic financial insights from client financial reports. It focuses on visualizing key financial trends, including Net Worth, Revenue, and Profit After Tax.

**Tool used:** Python (pandas; matplotlib)

**How It Works**
The script automates the process of turning raw financial data into an easily understandable visual report.

*Data Ingestion:* It reads financial data provided by the client, typically from a structured file format.

*Data Processing:* It uses pandas to clean and prepare the data, ensuring it's ready for analysis.

*Calculations:* Key financial metrics such as Net Worth, Revenue, and Profit After Tax are calculated based on the input data.

*Visualization:* Using matplotlib, the script generates a series of line plots. Each plot represents a different financial metric, allowing for a clear, at-a-glance view of the client's financial performance over time.


**Visual Output**
*Net Worth Trend:* A line plot showing how the client's net worth has changed over a specified period. This helps visualize growth or decline in overall financial health.

*Revenue Trend:* A line plot illustrating the company's total income over time. This metric is crucial for understanding sales performance and market demand.

*Profit After Tax Trend:* A line plot that displays the final profit after all expenses and taxes are accounted for. This is a key indicator of profitability and operational efficiency.

```python
file_path = path + 'CLIENT_FINACIAL_20250513.xlsx'

xls = pd.ExcelFile(file_path)
sheet_names = xls.sheet_names
RA = sheet_names[-1]
PL = sheet_names[-4]
BS = sheet_names[-5]
```

```python
print(BS)
```

```python
df = pd.read_excel(file_path , sheet_name= RA , index_col = 4, skiprows = 6)
```

```python
print(df.iloc[10,-1])
print(df.iloc[14,-1])
print(df.iloc[17,-1])
```

```python
insight = []
CR = df.iloc[10,-1]
LV = df.iloc[14,-1]
CP = df.iloc[17,-1]

if CR >= 1.1 and CR < 1.5:
    insight.append("Current Ratios is 2")
elif CR >= 1.5:
    insight.append("Current Ratios is 1")
elif CR >= 0.8 and CR < 1.5:
    insight.append("Current Ratios is 3")
elif CR < 0.8:
    insight.append("Current Ratios is 4")
else:
    insight.append("Missing Value")

if LV < 50 and LV != 0:
    insight.append("Leverage is 1")
elif LV < 60 and LV > 50:
    insight.append("Leverage is 2")
elif LV < 80 and LV > 70:
    insight.append("Leverage is 3")
elif LV > 80:
    insight.append("Leverage is 4")
else:
    insight.append("Missing Value")

if CP < 30:
    insight.append("Collection Days is 1")
elif CP < 60 and CP > 30:
    insight.append("Collection Days is 2")
elif CP > 60:
    insight.append("Collection Days is 3 or 4")
else:
    insight.append("Missing Value")
    
   
print(insight)

  
    
```
